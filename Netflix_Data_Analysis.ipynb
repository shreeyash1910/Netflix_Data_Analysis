```python
# ==========================================
# Import Required Libraries
# ==========================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")

plt.style.use("ggplot")
sns.set_theme(style="whitegrid")

pd.set_option("display.max_columns",None)

print("Libraries Imported Successfully")
```

```python
#  Load Dataset
df = pd.read_csv("netflix_titles.csv")   

print("Dataset Loaded Successfully")
```

```python
#   Dataset Overview

print("="*70)

print("Shape of Dataset")
print(df.shape)

print("="*70)

print("Column Names")
print(df.columns.tolist())

print("="*70)

print("Data Types")
print(df.dtypes)

print("="*70)

print("First 5 Rows")
display(df.head())

print("="*70)

print("Last 5 Rows")
display(df.tail())
```

```python
#   Dataset Information
df.info()
```

```python
#  Statistical Summary
display(df.describe(include="all"))
```

```python
#   Check Missing Value

missing=df.isnull().sum()

missing_percent=(missing/len(df))*100

missing_table=pd.DataFrame({

"Missing Values":missing,

"Percentage":missing_percent.round(2)

})

display(missing_table.sort_values("Percentage",ascending=False))
```

```python
#  Missing Value Visualization
plt.figure(figsize=(12,6))

sns.barplot(

x=missing_table["Percentage"],

y=missing_table.index,

palette="viridis"

)

plt.title("Missing Values Percentage",fontsize=18)

plt.xlabel("Percentage")

plt.ylabel("Columns")

plt.show()
```

```python
#   Duplicate Values
print("Duplicate Rows :",df.duplicated().sum())
```

```python
#   Check Null Values Again
df.isnull().sum()
```

```python
# Visualizations  (Correlation Heatmap)

plt.figure(figsize=(10,8))

sns.heatmap(

df.select_dtypes(include=np.number).corr(),

annot=True,

cmap="coolwarm",

linewidth=1

)

plt.title("Correlation Heatmap",fontsize=18)

plt.show()
```

```python
#   Histogram

df.hist(

figsize=(14,10),

bins=20,

edgecolor="black"

)

plt.tight_layout()

plt.show()
```

```python
#  Boxplots
numeric=df.select_dtypes(include=np.number)

for col in numeric.columns:

    plt.figure(figsize=(8,4))

    sns.boxplot(x=df[col])

    plt.title(col)

    plt.show()
```

```python
#  Countplot for Categorical Columns

categorical=df.select_dtypes(include="object")

for col in categorical.columns[:5]:

    plt.figure(figsize=(12,5))

    sns.countplot(

        y=df[col],

        order=df[col].value_counts().head(10).index,

        palette="Set2"

    )

    plt.title(col)

    plt.show()
```

```python
#  Pie Chart

if len(categorical.columns)>0:

    column=categorical.columns[0]

    df[column].value_counts().head(5).plot(

        kind="pie",

        autopct="%1.1f%%",

        figsize=(6,6),

        shadow=True

    )

    plt.title(column)

    plt.ylabel("")

    plt.show()
```

```python
#  Pairplot

sns.pairplot(

df.select_dtypes(include=np.number),

diag_kind="hist"

)

plt.show()
```
